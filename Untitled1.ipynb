```python
def closest_to_zero(numbers):
    closest = numbers[0]
    for num in numbers:
        if abs(num) < abs(closest) or (abs(num) == abs(closest) and num > closest):
            closest = num
    return closest

numbers = [1, 2, 4, 6, 8]
result = closest_to_zero(numbers)
print(f"The number closest to zero is: {result}")
```
